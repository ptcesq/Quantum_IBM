## Data 

```python
fileIn = "C:\\Users\\patri\\OneDrive\\Projects\\Quantum_IBM\\R_v_M\\data\\iris.csv"
columns = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width','target']
df = pd.read_csv(fileIn, names=columns)
df.head()
```

```python
set(list(df['target']))
```

```python
def fix_name(target_in): 
    if target_in == 'Iris-setosa': 
        target_out = 0
    elif target_in == 'Iris-versicolor':
        target_out = 1 
    else: 
        target_out = 2 
    return target_out 
```

```python
# transform target 
df['target'] = df['target'].apply(fix_name)
```
