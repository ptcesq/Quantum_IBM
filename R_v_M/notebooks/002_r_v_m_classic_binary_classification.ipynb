```python
# Import Libraries 

import pandas as pd 
import numpy as np 

# train test split 
from sklearn.model_selection import train_test_split

# model libraries 
from sklearn.ensemble import RandomForestClassifier
```

```python
# set the seed 
model_seed = 122
```

## Data 

```python
fileIn = "C:\\Users\\patri\\OneDrive\\Projects\\Quantum_IBM\\R_v_M\\data\\sonar.all-data.csv"
df = pd.read_csv(fileIn)
df.head()
```

```python
# find shape
df.shape
```

```python
# get the X,y vectors 
X = df.iloc[:, :60]
y = df.iloc[:, -1]
```

```python
# check results 
print("X: ", X.shape)
print("y: ", y.shape)
```

```python
# train/test split 
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=model_seed)
```

```python
# validate split 
print("Shape of X_train:", X_train.shape)
print("Shape of X_test:", X_test.shape)
print("Shape of y_train:", y_train.shape)
print("Shape of y_test:", y_test.shape)
```

## Model 

```python
# Model Evaluation 

def evaluate_model(model, x_test, y_test):
    from sklearn import metrics

    # Predict Test Data 
    y_pred = model.predict(x_test)

    # Calculate accuracy, precision, recall, f1-score, and kappa score
    acc = metrics.accuracy_score(y_test, y_pred) * 100
    prec = metrics.precision_score(y_test, y_pred) * 100
    rec = metrics.recall_score(y_test, y_pred) * 100
    f1 = metrics.f1_score(y_test, y_pred)
    kappa = metrics.cohen_kappa_score(y_test, y_pred)

    # Calculate area under curve (AUC)
    y_pred_proba = model.predict_proba(x_test)[::,1]
    fpr, tpr, _ = metrics.roc_curve(y_test, y_pred_proba)
    auc = metrics.roc_auc_score(y_test, y_pred_proba) * 100 

    # Display confussion matrix
    cm = metrics.confusion_matrix(y_test, y_pred)

    return {'acc': acc, 'prec': prec, 'rec': rec, 'f1': f1, 'kappa': kappa, 
            'fpr': fpr, 'tpr': tpr, 'auc': auc, 'cm': cm}
```

```python
# Building Random Forest model 
rf = RandomForestClassifier(random_state=0)
rf.fit(X_train, y_train)
```

```python
# Evaluate Model
rf_eval = evaluate_model(rf, X_test, y_test)

# Print result
print('Accuracy:', rf_eval['acc'])
print('Precision:', rf_eval['prec'])
print('Recall:', rf_eval['rec'])
print('F1 Score:', rf_eval['f1'])
print('Cohens Kappa Score:', rf_eval['kappa'])
print('Area Under Curve:', rf_eval['auc'])
print('Confusion Matrix:\n', rf_eval['cm'])
```
