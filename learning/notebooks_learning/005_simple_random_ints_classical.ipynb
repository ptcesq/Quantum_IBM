## Generate Random numbers using classical techniques

```python
import numpy as np
```

```python
num_array = np.random.randint(1, 266, size=1500)
```

```python
import matplotlib.pyplot as plt
plt.hist(num_array, bins=256)
plt.title('Histogram of Random Numbers genterated classically')
plt.xlabel('Number')
plt.ylabel('Frequency')
plt.xlim(0, 265)
plt.show()
```
