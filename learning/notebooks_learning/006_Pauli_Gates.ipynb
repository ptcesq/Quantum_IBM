# Matrix Multiplcation 


```python
import numpy as np 
```

```python
A = np.array([[1, 0], [0, -1]])
v = np.array([[1,0]]).reshape(2, 1)

C = np.dot(A, v)

print("A: ", A)
print("v: ", v)
print("C: ", C)
```

```python
A = np.array([[1, 0], [0, -1]])
v = np.array([[0, 1]]).reshape(2, 1)

C = np.dot(A, v)

print("A: ", A)
print("v: ", v)
print("C: ", C)
```
