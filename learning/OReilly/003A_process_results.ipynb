## Post process quantum output 

```python
import pandas as pd
```

```python
fileIn = "C:\\Users\\patri\\OneDrive\\Projects\\Quantum_IBM\\learning\\OReilly\\data\\random_nums.csv"
df = pd.read_csv(fileIn, header=None, names=['value'])
df['value'] = df['value'].astype(str)
df['value'] = df['value'].str.zfill(8)
df.head()
```

```python
int_array = []

for i in range(0, len(df)-1):
	t1 = df.iloc[i, 0]
	t2 = int(t1, 2)
	int_array.append(t2)
```

```python
int_array
```

```python
import matplotlib.pyplot as plt
plt.hist(int_array, bins=256, density=True)
plt.xlabel('Value')
plt.ylabel('Density')
plt.title('Histogram of Random Numbers Generated by QPU')
plt.show()
```
