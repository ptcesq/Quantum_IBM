# Simple Random Byte Generator

## Libraries

```python
# for the quantum circuit
from qiskit.circuit import QuantumCircuit
# for the backend
from qiskit_aer import AerSimulator
# for the transpiler
from qiskit import transpile



print("Llibraries Loaded")
```

## Backend

```python
# select the backend
backend = AerSimulator()

print(f"Using backend: {backend}")
print("Backend Loaded")
```

## Circuit

```python

n =8
qc = QuantumCircuit(n)
for i in range(n):
    qc.h(i)
qc.measure_all()

qc.draw('mpl')
```

```python
# compile the circuit
compiled_circuit = transpile(qc, backend=backend)
```

```python
results = backend.run(compiled_circuit).result()
```

```python
# get the results 
counts = results.get_counts(compiled_circuit)
print("Results:")   
print(counts)
```

```python
import matplotlib.pyplot as plt
# plot the results  
plt.bar(counts.keys(), counts.values())
plt.xlabel('States')    
plt.show()
```
