# Three Qbit Entanglement (GHZ)

## 1. Load Libraries

```python
# for the quantum circuit
from qiskit.circuit import QuantumCircuit
# for the backend
from qiskit_aer import AerSimulator
# for the transpiler
from qiskit import transpile



print("Llibraries Loaded")
```

## Backend

```python
# select the backend
backend = AerSimulator()
print(f"Using backend: {backend}")
print("Backend Loaded")
```

## Circuit

```python

n = 8   # number of qubits
# create a quantum circuit with n qubits
qc = QuantumCircuit(n)  
print(f"Quantum Circuit Created with {n} qubits")

qc.h(0)  # apply Hadamard gate to qubit 0
qc.cx(0, 1)  # apply CNOT gate with control qubit 0 and target qubit 1
qc.cx(0, 2)  # apply CNOT gate with control qubit 0 and target qubit 2
qc.measure_all()  # measure all qubits

qc.draw('mpl')
```

```python
# compile the circuit
compiled_circuit = transpile(qc, backend=backend)
```

```python
compiled_circuit.draw('mpl')
```

```python
results = backend.run(compiled_circuit, shots=1500).result()
```

```python
# get the results 
counts = results.get_counts(compiled_circuit)
print("Results:")   
print(counts)
```

```python
import matplotlib.pyplot as plt
# plot the results  
plt.bar(counts.keys(), counts.values())
plt.xlabel('States')    
plt.show()
```
