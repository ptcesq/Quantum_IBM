# Update Credentials 

```python
import tkinter as tk
from tkinter import simpledialog

root = tk.Tk()
root.withdraw()  # Hide the main window

user_input = simpledialog.askstring("Input", "Please enter your new token:")

if user_input:
    print(f"Token Recieved: -{user_input}-")
else:
    print("No input provided.")
```

```python
from qiskit_ibm_runtime import QiskitRuntimeService
service = QiskitRuntimeService.save_account(
    token=user_input,                           # Your confidential token.
    instance="ptc_work",                        # Optionally specify the instance to use.
    name="ptc_work",                            # Optionally name this set of account credentials.
    set_as_default=True,                        # Optionally set these as your default credentials.
    )
```
