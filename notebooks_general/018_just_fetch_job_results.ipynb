# Fetch Job

```python
# Get api token 
import keyring as k3

# Get the API token from keyring
api_token = k3.get_password('ibm_quantum', 'api_token')

import datetime
```

```python
from qiskit_ibm_runtime import QiskitRuntimeService

service = QiskitRuntimeService(
    channel='ibm_quantum',
    instance='ibm-q/open/main',
    token= api_token
)
job = service.job('d0sm2sbvx7bg00819bk0')
job_result = job.result()

# To get counts for a particular pub result, use 
#
pub_result = job_result
# where <idx> is the index of the pub and <classical register> is the name of the classical register. 
# You can use circuit.cregs to find the name of the classical registers.
```

```python
print(job_result)
```

```python
print(job_result.metadata) # Access global metadata
```

```python
# get the data from the pub object
measurement_result = job_result[0].data
```

```python
# get the array of bits from the measurement result
bit_array = measurement_result.meas
print("Shape of Array: ", bit_array.shape)
print("NUmber of Bits:", bit_array.num_bits)
print("Array: \n", bit_array.array)

```

```python
# put the data into a numpy array
import numpy as np
result_array = np.array(bit_array.array)
```
