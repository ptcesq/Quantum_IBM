## Just playing around with rustwokx

```python
import rustworkx as rx
import numpy as np
```

```python
G = rx.PyGraph()
```

```python
n = 5 
G.add_nodes_from(np.arange(0, n, 1))
```

```python
elist = [
    (0, 1, 1.0), 
    (0, 2, 1.0),
    (0, 4, 1.0),
    (1, 2, 1.0),
    (2, 3, 1.0),
    (3, 4, 1.0)] 
```

```python
# Add edges to the graph
G.add_edges_from(elist)
```

```python
from rustworkx.visualization import mpl_draw
mpl_draw(G,with_labels=True, edge_labels=str)  
```
